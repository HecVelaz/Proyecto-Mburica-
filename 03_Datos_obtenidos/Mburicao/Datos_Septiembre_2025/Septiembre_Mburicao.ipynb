```python
import pandas as pd

```

```python
from google.colab import drive
drive.mount('/content/drive')

ruta = "/content/drive/MyDrive/Colab Notebooks/IA/Mburicao/"

df0 = pd.read_csv(ruta + "Mburicao_Sil.csv",
                  header=None,
                  names=["fecha","nivel"])
```

```python
inicio = '2025-08-27 14:30:00'
fin = '2025-09-30 23:55:00'
mascara = (df0['fecha'] >= inicio) & (df0['fecha'] <= fin)
df_septiembre = df0.loc[mascara].copy()
```

```python
df_septiembre['fecha'] = pd.to_datetime(df_septiembre['fecha'])
df_septiembre= df_septiembre.sort_values('fecha').reset_index(drop=True)
```

```python

df_septiembre['fecha'] = df_septiembre['fecha'].dt.round('5min')
df_septiembre
```

```python
# =============================================
# ELIMINAR FILAS CON FECHA DUPLICADA (conservar la primera)
# =============================================

# Verificar cuántas filas tenemos antes
print(f"Filas antes de eliminar duplicados: {len(df_septiembre)}")

# Identificar fechas repetidas (opcional, solo para diagnóstico)
duplicados_fecha = df_septiembre.duplicated(subset=['fecha'], keep=False)
print(f"Filas con fecha duplicada (incluyendo primera aparición): {duplicados_fecha.sum()}")
if duplicados_fecha.sum() > 0:
    print("Ejemplo de fechas repetidas (primeras 5):")
    display(df_septiembre[duplicados_fecha].sort_values('fecha').head(10))

# Eliminar duplicados basados en la columna 'fecha', conservando la primera ocurrencia
df_septiembre = df_septiembre.drop_duplicates(subset=['fecha'], keep='first')

# Verificar después
print(f"Filas después de eliminar duplicados: {len(df_septiembre)}")
```

```python
# Contar cuántos duplicados quedan (debe ser 0)
print("Número de filas con fecha duplicada:", df_septiembre.duplicated(subset=['fecha']).sum())
```

```python
# =============================================
# Crear índice regular de 5 minutos y reindexar
# =============================================

# Asegurar que la columna 'fecha' sea el índice
df_septiembre = df_septiembre.set_index('fecha').sort_index()

# Definir rango completo
start_time = df_septiembre.index.min()
end_time = df_septiembre.index.max()
time_index = pd.date_range(start=start_time, end=end_time, freq='5min')

# Reindexar: las marcas faltantes quedarán con NaN
df_regular = df_septiembre.reindex(time_index)
df_regular.index.name = 'fecha'

# Verificar cuántos NaN se han introducido
print(f"Filas después de reindexar: {len(df_regular)}")
print(f"Valores nulos en 'nivel': {df_regular['nivel'].isna().sum()}")
```

```python
df_regular
```

```python
# Asegurar que el índice sea datetime y esté ordenado
df_regular = df_regular.sort_index()

fecha_inicio = pd.to_datetime('2025-09-12 10:10:00')

# Máscara para filas desde la fecha de inicio (inclusive)
desde_fecha = df_regular.index >= fecha_inicio

# Buscar el último valor no nulo antes de la fecha de inicio
ultimo_valor = df_regular.loc[df_regular.index < fecha_inicio, 'nivel'].last_valid_index()

if ultimo_valor is not None and desde_fecha.any():
    # Crear una serie que incluya el último valor antes y los datos desde la fecha
    # (esto garantiza que la interpolación tenga un punto de partida)
    serie_a_interpolar = pd.concat([
        df_regular.loc[[ultimo_valor], 'nivel'],          # último valor real antes
        df_regular.loc[desde_fecha, 'nivel']              # datos desde la fecha
    ])

    # Interpolar linealmente sobre esta serie
    serie_interpolada = serie_a_interpolar.interpolate(method='linear')

    # Asignar los valores interpolados (excluyendo el punto de anclaje) al DataFrame original
    df_regular.loc[desde_fecha, 'nivel'] = serie_interpolada.iloc[1:].values

    print("Interpolación aplicada desde", fecha_inicio)
else:
    print("No hay valores anteriores válidos o no hay datos desde la fecha indicada.")

# Verificar que no queden nulos después de la fecha (pueden quedar nulos antes si los había)
print("Nulos después de la fecha:", df_regular.loc[desde_fecha, 'nivel'].isna().sum())
```

```python
# =============================================
# APLICAR (7 - nivel)
# =============================================
df_regular['nivel'] = 7 - df_regular['nivel']
```

```python
# =============================================
# GRAFICAR INTERACTIVAMENTE CON PLOTLY
# =============================================
import plotly.express as px

fig = px.line(df_regular, x=df_regular.index, y='nivel',
              title='Nivel de agua (m) – septiembre 2025 (corregido +1h, interpolado)')
fig.update_layout(hovermode='x unified')
fig.show()
```
